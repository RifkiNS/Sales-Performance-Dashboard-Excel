```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
from google.colab import files
```

```python
# 1. Generate Master Produk
skus = [f"PROD{str(i).zfill(3)}" for i in range(1, 11)]
nama_produk = [
    "Laptop Aspire", "Smartphone X", "Meja Kerja Minimalis", "Kursi Ergonomis",
    "Kemeja Oxford", "Celana Chino", "Monitor 24 Inch", "Mouse Wireless",
    "Rak Buku Kayu", "Jaket Bomber"
]
kategori_list = ["Elektronik", "Elektronik", "Furnitur", "Furnitur", "Pakaian", "Pakaian", "Elektronik", "Elektronik", "Furnitur", "Pakaian"]
costs = [5000000, 3000000, 800000, 600000, 150000, 180000, 1200000, 150000, 400000, 200000]
prices = [7500000, 4500000, 1200000, 950000, 250000, 280000, 1800000, 250000, 650000, 350000]

df_produk = pd.DataFrame({
    'SKU': skus,
    'Nama Produk': nama_produk,
    'Kategori': kategori_list,
    'Cost': costs,
    'Price': prices
})
```

```python
# 2. Generate Master Target
bulan_list = ["Januari", "Februari", "Maret", "April", "Mei", "Juni", "Juli", "Agustus", "September", "Oktober", "November", "Desember"]
df_target = pd.DataFrame({
    'Kategori': ["Elektronik", "Furnitur", "Pakaian"],
    'Januari': [50000000, 20000000, 15000000],
    'Februari': [55000000, 22000000, 16500000],
    'Maret': [60000000, 24000000, 18000000],
    'April': [65000000, 26000000, 19500000],
    'Mei': [70000000, 28000000, 21000000],
    'Juni': [75000000, 30000000, 22500000],
    'Juli': [80000000, 32000000, 24000000],
    'Agustus': [85000000, 34000000, 25500000],
    'September': [90000000, 36000000, 27000000],
    'Oktober': [95000000, 38000000, 28500000],
    'November': [100000000, 40000000, 30000000],
    'Desember': [110000000, 45000000, 35000000]
})
```

```python
# 3. Generate 1000 Data Transaksi (Dengan format teks berantakan khas input admin manual)
np.random.seed(42)
random.seed(42)

start_date = datetime(2025, 1, 1)
end_date = datetime(2025, 12, 31)
dates = [start_date + timedelta(days=int(np.random.randint(0, 365))) for _ in range(1000)]
dates.sort()

nama_mentah = ["  budi santoso  ", "RINA marlina", "  siti RAHMA  ", "andi WIJAYA", "  DEWI lestari  ", "Eko Prasetyo", " mega utami ", "Bambang H.", "Ahmad Fauzi", "  Sari Yuliana  "]
wilayah_list = ["Pusat", "Barat", "Selatan", "Utara", "Timur"]

txn_id = [f"TXN-2025-{str(i).zfill(3)}" for i in range(1, 1001)]
pelanggan = [random.choice(nama_mentah) for _ in range(1000)] # Sengaja berantakan untuk simulasi TRIM & PROPER
wilayah = [random.choice(wilayah_list) for _ in range(1000)]
chosen_sku = [random.choice(skus) for _ in range(1000)]
jumlah_beli = [int(np.random.choice([1, 2, 3, 4, 5, 10], p=[0.4, 0.3, 0.15, 0.08, 0.05, 0.02])) for _ in range(1000)]

df_transaksi = pd.DataFrame({
    'ID Transaksi': txn_id,
    'Tanggal': [d.strftime('%Y-%m-%d') for d in dates],
    'Nama Pelanggan Raw': pelanggan, # Kolom untuk dibersihkan
    'Wilayah': wilayah,
    'SKU': chosen_sku,
    'Jumlah Beli': jumlah_beli
})
```

```python
# Menyiapkan kolom kosong untuk formula Excel Anda
df_transaksi['Nama Pelanggan Clean'] = ""
df_transaksi['Kategori'] = ""
df_transaksi['Harga Jual'] = ""
df_transaksi['Total Revenue'] = ""
df_transaksi['Target Bulanan Kategori'] = ""

# Reorder kolom agar rapi saat dibuka di Excel
df_transaksi = df_transaksi[['ID Transaksi', 'Tanggal', 'Nama Pelanggan Raw', 'Nama Pelanggan Clean', 'Wilayah', 'SKU', 'Jumlah Beli', 'Kategori', 'Harga Jual', 'Total Revenue', 'Target Bulanan Kategori']]
```

```python
# 4. Eksport ke Excel dengan beberapa Sheet sekaligus
filename = 'Sales_Operations_Analytics_Portfolio.xlsx'
with pd.ExcelWriter(filename, engine='xlsxwriter') as writer:
    df_transaksi.to_excel(writer, sheet_name='Data_Transaksi', index=False)
    df_produk.to_excel(writer, sheet_name='Master_Produk', index=False)
    df_target.to_excel(writer, sheet_name='Master_Target', index=False)

    # Sedikit formatting dasar agar header menarik
    workbook  = writer.book
    header_format = workbook.add_format({'bold': True, 'font_color': 'white', 'fg_color': '#1F4E78', 'border': 1})

    for sheet in writer.sheets.values():
        sheet.set_row(0, 20, header_format)
        sheet.autofit()
```

```python
# 5. Otomatis download file di Google Colab
files.download(filename)
print("File berhasil di-generate! Cek folder download pada peramban/browser Anda.")
```
